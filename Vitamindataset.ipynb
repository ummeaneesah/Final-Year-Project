```python
#Load new dataset and openpyxl library to be able to read the excel file
data_frame = pd.read_excel('/content/Vitamin_Dataset_Cleaned.xlsx', engine='openpyxl')
#Removing white space and characters in columns c.strip for c (for all the columns in the dataset)
data_frame.columns = [c.strip() for c in data_frame.columns]
```

```python
#Convert written numbers into numeric values by creating a dictionary everything inside of the {} assigns keys to the words giving a value to the corresponding interger like one is 1
word_to_num = {
    'zero':0,'one':1,'two':2,'three':3,'four':4,'five':5,'six':6,'seven':7,'eight':8,'nine':9,'ten':10,
    'eleven':11,'twelve':12,'thirteen':13,'fourteen':14,'fifteen':15,'sixteen':16,'seventeen':17,'eighteen':18,'nineteen':19,'twenty':20
}
```

```python
#Create dictionary assigning words to numbers
freq_words = {
    'once':1,'one time':1,'one-time':1,
    'twice':2,'two times':2,
    'three times':3,
    'four times':4,
    'five times':5,
    'six times':6,
}
```

```python
#Assigning types of supplements to columns
data_frame['is_botanical'] = data_frame['Product Type'].str.contains('botanical')
data_frame['is_vitamin'] = data_frame['Product Type'].str.contains('vitamin')
data_frame['is_mineral'] = data_frame['Product Type'].str.contains('mineral')
#bitwise| for other combination or multi
data_frame['is_combination'] = data_frame['Product Type'].str.contains('other combinations') | data_frame['Product Type'].str.contains('multi')
```

```python
#Assigning forms of supplements to columns
form = data_frame['Supplement Form'].fillna('').str.lower()
data_frame['is_liquid'] = form.str.contains('liquid')
data_frame['is_capsule'] = form.str.contains('capsule')
data_frame['is_tablet'] = form.str.contains('tablet|caplet|lozenge|wafer')
#bitwise | for different forms if not these forms ignore
fd_powder_re = re.compile(r'powder|granules|crystals', re.IGNORECASE)
#if the text value lambda is true or false
data_frame['is_powder'] = form.apply(lambda x: bool(fd_powder_re.search(x)) if isinstance(x, str) else False)
```

```python
#Replacing missing values with an empty string and converting all words to lowercase
text = data_frame['Suggested Use'].fillna('').str.lower()
#Assigning these values to a column
data_frame['mentions_bedtime'] = text.str.contains('bedtime')
data_frame['mentions_caffeine'] = text.str.contains('caffeine')
data_frame['mentions_empty_stomach'] = text.str.contains('empty stomach')
data_frame['mentions_shake_well'] = text.str.contains('shake well')
data_frame['mentions_refrigerate'] = text.str.contains('refrigerate')
#Calculates how complex direction to take supplements are taking into account all these factors axis 1 means to run data row by row
data_frame['direction_complexity_score'] = (
    data_frame[['mentions_bedtime','mentions_caffeine','mentions_empty_stomach','mentions_shake_well','mentions_refrigerate']]
    .sum(axis=1)
)
```

```python
#True or False list for all these types of supplements
def make_label(row):
    labels = []
    if row.get('is_vitamin'): labels.append('Vitamin')
    if row.get('is_mineral'): labels.append('Mineral')
    if row.get('is_botanical'): labels.append('Botanical')
    if row.get('is_combination'): labels.append('Combo')
    #if it is multiple its combined with a /
    form_bits = []
    if row.get('is_liquid'): form_bits.append('Liquid')
    if row.get('is_capsule'): form_bits.append('Capsule')
    if row.get('is_tablet'): form_bits.append('Tablet/Caplet/Lozenge')
    if row.get('is_powder'): form_bits.append('Powder')
    if form_bits:
        labels.append('/'.join(form_bits))
    if row.get('mentions_bedtime'): labels.append('Bedtime')
    if row.get('mentions_caffeine'): labels.append('Caffeine')
    if row.get('mentions_empty_stomach'): labels.append('Empty stomach')
    if row.get('mentions_shake_well'): labels.append('Shake well')
    if row.get('mentions_refrigerate'): labels.append('Refrigerate')
    #seperated with commas
    return ', '.join(labels)
```

```python
#Creating a rule label to be applied to every row
data_frame['rule_label'] = data_frame.apply(make_label, axis=1)

preview = data_frame[['Product Name','Brand Name','Serving Size','Suggested Use','rule_label','direction_complexity_score']].copy()
```

```python
output_excel = 'Vitamin_RuleModel_Output.xlsx'

# Ensure preview_sorted is defined before saving if it's not already.
# If it is already defined from a previous execution, this will just re-sort.
if 'preview' in globals():
    preview_sorted = preview.sort_values(by='direction_complexity_score', ascending=False)
else:
    print("Warning: 'preview' DataFrame not found. Cannot create 'preview_sorted'.")
    preview_sorted = pd.DataFrame() # Create an empty DataFrame to avoid errors


with pd.ExcelWriter(output_excel, engine='openpyxl') as writer:
    data_frame.to_excel(writer, sheet_name='enriched', index=False)
    if not preview_sorted.empty:
        preview_sorted.to_excel(writer, sheet_name='top_daily_units', index=False)

print(f"All changes saved to {output_excel}")
```
